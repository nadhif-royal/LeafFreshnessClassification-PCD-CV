## 2. Pipeline PCD v2, Ekstraksi 28 Fitur, QC Segmentasi, Mesin Inferensi, & Dashboard
Fungsi pada bagian ini disalin dari Notebook 02, 03, dan 04. Kesesuaiannya **diverifikasi oleh sel uji paritas (Bagian 2b)**, bukan sekadar diasumsikan:
* **PCD v2:** Resize proporsional sisi panjang 800 px, Gamma Adaptif `L**gamma` menuju target median `L* = 0,716` (`[0,8; 1,25]`), Bilateral Filter (`d=9, sigma=75`), CLAHE pada kanal `L*` (`clipLimit=2,0`), Dual Otsu (`b* OR S`) + Morfologi Open/Close (`7x7`) + Largest Connected Component, serta *letterbox ROI crop* `224x224` dengan padding 8%.
* **Quality Control (QC) Segmentasi:** Foto ditandai tidak andal apabila luas area daun di luar rentang kalibrasi atau mask menyentuh >25% tepi bingkai foto (indikasi latar bukan kertas putih).
* **Ekstraksi 28 Fitur:** 12 fitur statistik warna HSV dan CIE-LAB khusus piksel daun, 6 fitur GLCM bebas latar (intensitas daun digeser ke `[1..255]`), serta 10 bin histogram Uniform LBP (`P=8, R=1`).
* **Inferensi & Dashboard 5 Panel:** Menampilkan citra mentah, mask Otsu, C2 ROI + kotak deteksi YOLOv8n, Grad-CAM MobileNetV2 (dengan *caching* grafik), serta grafik perbandingan *confidence*, latensi *end-to-end* (ms), dan **Konsensus Multi-Model**.

```python
# === BAGIAN 2: PIPELINE PCD V2, QC SEGMENTASI, 28 FITUR, INFERENSI, & DASHBOARD ===
WORK_LONG = 800
TARGET_L  = 0.716

QC_AREA_LO, QC_AREA_HI = 2.0, 75.0   # Ambang awal; kalibrasi dari rentang area pada Bagian 2b
QC_BORDER_MAX = 0.25                 # >25% tepi foto tersentuh mask = segmentasi diduga gagal

COLOR_COLS = [
    'H_mean', 'H_std', 'S_mean', 'S_std', 'V_mean', 'V_std',
    'L_mean', 'L_std', 'a_mean', 'a_std', 'b_mean', 'b_std'
]
GLCM_PROPS = ['contrast', 'dissimilarity', 'homogeneity', 'energy', 'correlation', 'ASM']
GLCM_COLS  = [f'GLCM_{p}' for p in GLCM_PROPS]
LBP_COLS   = [f'LBP_bin_{i}' for i in range(10)]
ALL_FEATURE_COLS = COLOR_COLS + GLCM_COLS + LBP_COLS

# === A. FUNGSI PIPELINE PCD V2 + QUALITY CONTROL ===
def resize_long(img, long_side=WORK_LONG):
    h, w = img.shape[:2]
    s = long_side / max(h, w)
    return cv2.resize(img, (round(w * s), round(h * s)), interpolation=cv2.INTER_AREA)

def adaptive_gamma_lab(img_bgr, target=TARGET_L, lo=0.8, hi=1.25):
    lab = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB)
    m = float(np.clip(lab[:, :, 0].mean() / 255.0, 0.05, 0.95))
    g = float(np.clip(np.log(target) / np.log(m), lo, hi))
    lut = (((np.arange(256) / 255.0) ** g) * 255).astype(np.uint8)
    lab[:, :, 0] = cv2.LUT(lab[:, :, 0], lut)
    return cv2.cvtColor(lab, cv2.COLOR_LAB2BGR), g

def apply_bilateral(img_bgr, d=9, sigma_color=75, sigma_space=75):
    return cv2.bilateralFilter(img_bgr, d=d, sigmaColor=sigma_color, sigmaSpace=sigma_space)

def apply_clahe_lab(img_bgr, clip_limit=2.0, tile_size=(8, 8)):
    lab = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)
    clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_size)
    l_clahe = clahe.apply(l)
    return cv2.cvtColor(cv2.merge([l_clahe, a, b]), cv2.COLOR_LAB2BGR)

def segment_leaf_otsu_morph(img_bgr):
    lab = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB)
    hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
    _, _, b = cv2.split(lab)
    _, s, _ = cv2.split(hsv)
    _, mask_b = cv2.threshold(b, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    _, mask_s = cv2.threshold(s, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    mask_raw = cv2.bitwise_or(mask_b, mask_s)
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7, 7))
    mask_clean = cv2.morphologyEx(mask_raw, cv2.MORPH_OPEN, kernel, iterations=1)
    mask_clean = cv2.morphologyEx(mask_clean, cv2.MORPH_CLOSE, kernel, iterations=2)
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(mask_clean, connectivity=8)
    if num_labels > 1:
        largest_label = 1 + np.argmax(stats[1:, cv2.CC_STAT_AREA])
        mask_clean = np.where(labels == largest_label, 255, 0).astype(np.uint8)
    return mask_raw, mask_clean

def crop_roi_with_padding(img_bgr, mask, target_size=(224, 224), pad_ratio=0.08):
    ys, xs = np.where(mask > 0)
    if len(xs) == 0 or len(ys) == 0:
        return cv2.resize(img_bgr, target_size), cv2.resize(mask, target_size, interpolation=cv2.INTER_NEAREST)
    x1, x2 = np.min(xs), np.max(xs)
    y1, y2 = np.min(ys), np.max(ys)
    h_box, w_box = y2 - y1, x2 - x1
    max_side = int(max(h_box, w_box) * (1.0 + 2 * pad_ratio))
    cx, cy = (x1 + x2) // 2, (y1 + y2) // 2
    half = max_side // 2
    img_masked = cv2.bitwise_and(img_bgr, img_bgr, mask=mask)
    h_img, w_img = img_bgr.shape[:2]
    canvas_img = np.zeros((max_side, max_side, 3), dtype=np.uint8)
    canvas_mask = np.zeros((max_side, max_side), dtype=np.uint8)
    src_x1, src_y1 = max(0, cx - half), max(0, cy - half)
    src_x2, src_y2 = min(w_img, cx - half + max_side), min(h_img, cy - half + max_side)
    dst_x1, dst_y1 = src_x1 - (cx - half), src_y1 - (cy - half)
    dst_x2, dst_y2 = dst_x1 + (src_x2 - src_x1), dst_y1 + (src_y2 - src_y1)
    canvas_img[dst_y1:dst_y2, dst_x1:dst_x2] = img_masked[src_y1:src_y2, src_x1:src_x2]
    canvas_mask[dst_y1:dst_y2, dst_x1:dst_x2] = mask[src_y1:src_y2, src_x1:src_x2]
    roi_resized = cv2.resize(canvas_img, target_size, interpolation=cv2.INTER_AREA)
    mask_resized = cv2.resize(canvas_mask, target_size, interpolation=cv2.INTER_NEAREST)
    roi_resized = cv2.bitwise_and(roi_resized, roi_resized, mask=mask_resized)
    return roi_resized, mask_resized

def run_pcd_v2(img_bgr):
    t0 = time.perf_counter()
    im_work = resize_long(img_bgr, WORK_LONG)
    im_gamma, gamma_val = adaptive_gamma_lab(im_work, TARGET_L)
    im_c1 = apply_clahe_lab(apply_bilateral(im_gamma), clip_limit=2.0)
    _, mask_full = segment_leaf_otsu_morph(im_c1)
    c2_roi, mask_roi = crop_roi_with_padding(im_c1, mask_full, target_size=(224, 224))
    pcd_ms = (time.perf_counter() - t0) * 1000.0
    leaf_area_pct = float((mask_full > 0).mean() * 100.0)

    border = np.concatenate([mask_full[0, :], mask_full[-1, :], mask_full[:, 0], mask_full[:, -1]])
    border_frac = float((border > 0).mean())
    qc_msgs = []
    if leaf_area_pct < QC_AREA_LO:
        qc_msgs.append(f"area daun {leaf_area_pct:.1f}% terlalu kecil")
    if leaf_area_pct > QC_AREA_HI:
        qc_msgs.append(f"area daun {leaf_area_pct:.1f}% terlalu besar")
    if border_frac > QC_BORDER_MAX:
        qc_msgs.append(f"{border_frac * 100:.0f}% tepi foto tersentuh mask")

    return {
        'im_work': im_work,
        'im_c1': im_c1,
        'mask_full': mask_full,
        'c2_roi': c2_roi,
        'mask_roi': mask_roi,
        'gamma_val': gamma_val,
        'leaf_area_pct': leaf_area_pct,
        'pcd_ms': pcd_ms,
        'qc_ok': len(qc_msgs) == 0,
        'qc_msg': '; '.join(qc_msgs) if qc_msgs else 'OK'
    }

# === B. EKSTRAKSI 28 FITUR HANDCRAFTED (SINKRON NOTEBOOK 03) ===
def extract_handcrafted_features(img_bgr, mask=None, use_mask=True):
    hsv  = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV).astype(np.float32)
    lab  = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB).astype(np.float32)
    gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    valid = (mask > 0) if (use_mask and mask is not None and np.sum(mask > 0) > 50) else np.ones(gray.shape, dtype=bool)

    feats = {}
    for idx, ch_name in enumerate(['H', 'S', 'V']):
        vals = hsv[:, :, idx][valid]
        feats[f'{ch_name}_mean'] = float(np.mean(vals))
        feats[f'{ch_name}_std']  = float(np.std(vals))
    for idx, ch_name in enumerate(['L', 'a', 'b']):
        vals = lab[:, :, idx][valid]
        feats[f'{ch_name}_mean'] = float(np.mean(vals))
        feats[f'{ch_name}_std']  = float(np.std(vals))

    if use_mask and mask is not None and np.sum(mask > 0) > 50:
        gray_glcm = np.where(valid, np.clip(gray, 1, 255), 0).astype(np.uint8)
        glcm = graycomatrix(
            gray_glcm, distances=[1, 2],
            angles=[0, np.pi/4, np.pi/2, 3*np.pi/4],
            levels=256, symmetric=True, normed=False
        )
        glcm[0, :, :, :] = 0
        glcm[:, 0, :, :] = 0
        glcm_sum = np.sum(glcm, axis=(0, 1), keepdims=True)
        glcm_norm = np.divide(glcm.astype(np.float64), glcm_sum, out=np.zeros_like(glcm, dtype=np.float64), where=glcm_sum > 0)
    else:
        glcm_norm = graycomatrix(
            gray, distances=[1, 2],
            angles=[0, np.pi/4, np.pi/2, 3*np.pi/4],
            levels=256, symmetric=True, normed=True
        )
    for prop in GLCM_PROPS:
        feats[f'GLCM_{prop}'] = float(np.mean(graycoprops(glcm_norm, prop)))

    lbp = local_binary_pattern(gray, P=8, R=1, method='uniform')
    hist, _ = np.histogram(lbp[valid], bins=np.arange(0, 11), density=True)
    for i in range(10):
        feats[f'LBP_bin_{i}'] = float(hist[i])
    return feats

# === C. GRAD-CAM MOBILENETV2 (DENGAN CACHING GRAFIK) ===
_GRADCAM_CACHE = {}

def compute_gradcam_overlay(c2_bgr, model, last_conv_layer_name='out_relu', alpha=0.45):
    img_rgb = cv2.cvtColor(c2_bgr, cv2.COLOR_BGR2RGB).astype(np.float32)
    img_tensor = np.expand_dims(mnv2_preprocess(img_rgb), axis=0)
    key = (id(model), last_conv_layer_name)
    if key not in _GRADCAM_CACHE:
        _GRADCAM_CACHE[key] = keras_models.Model(
            inputs=model.inputs,
            outputs=[model.get_layer(last_conv_layer_name).output, model.output]
        )
    grad_model = _GRADCAM_CACHE[key]
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_tensor)
        pred_index = tf.argmax(predictions[0])
        class_channel = predictions[:, pred_index]
    grads = tape.gradient(class_channel, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    heatmap = tf.squeeze(conv_outputs[0] @ pooled_grads[..., tf.newaxis]).numpy()
    heatmap = np.maximum(heatmap, 0)
    if np.max(heatmap) > 0:
        heatmap = heatmap / np.max(heatmap)
    heatmap_resized = cv2.resize(heatmap, (c2_bgr.shape[1], c2_bgr.shape[0]))
    heatmap_color = cv2.applyColorMap(np.uint8(255 * heatmap_resized), cv2.COLORMAP_JET)
    superimposed = cv2.addWeighted(c2_bgr, 1 - alpha, heatmap_color, alpha, 0)
    return cv2.cvtColor(superimposed, cv2.COLOR_BGR2RGB)

# === D. MESIN INFERENSI MULTI-MODEL ===
def run_multimodel_inference(img_bgr, filename='sample.jpg', true_label=None):
    pcd = run_pcd_v2(img_bgr)
    c2_roi   = pcd['c2_roi']
    mask_roi = pcd['mask_roi']
    preds = {}

    # 1. Prediksi Classical ML (28 fitur, aman untuk dict maupun Pipeline langsung)
    if LOADED_MODELS['ml'] is not None:
        t0 = time.perf_counter()
        bundle = LOADED_MODELS['ml']
        feats = extract_handcrafted_features(c2_roi, mask=mask_roi, use_mask=True)
        if isinstance(bundle, dict):
            feat_cols = bundle.get('feature_cols', ALL_FEATURE_COLS)
            pipe = bundle['pipeline']
        else:
            feat_cols, pipe = ALL_FEATURE_COLS, bundle
        x_in = np.array([[feats[c] for c in feat_cols]], dtype=np.float64)
        if hasattr(pipe, 'predict_proba'):
            probs_raw = pipe.predict_proba(x_in)[0]
            cls_order = list(pipe.classes_)
            prob_dict = {c: float(probs_raw[cls_order.index(c)]) for c in CLASS_NAMES if c in cls_order}
            pred_lbl = max(prob_dict, key=prob_dict.get)
            conf_ml = prob_dict[pred_lbl] * 100.0
        else:
            pred_lbl = str(pipe.predict(x_in)[0])
            prob_dict = {pred_lbl: 1.0}
            conf_ml = 100.0
        ml_ms = (time.perf_counter() - t0) * 1000.0
        preds['ML_Klasik'] = {
            'label': pred_lbl,
            'conf': conf_ml,
            'probs': prob_dict,
            'ms': ml_ms
        }

    # 2. Prediksi MobileNetV2 C2 + Grad-CAM
    gradcam_rgb = None
    if LOADED_MODELS['mobilenetv2'] is not None:
        t0 = time.perf_counter()
        img_rgb = cv2.cvtColor(c2_roi, cv2.COLOR_BGR2RGB).astype(np.float32)
        x_dl = np.expand_dims(mnv2_preprocess(img_rgb), axis=0)
        probs_dl = LOADED_MODELS['mobilenetv2'].predict(x_dl, verbose=0)[0]
        mnv2_ms = (time.perf_counter() - t0) * 1000.0
        prob_dict = {CLASS_NAMES[i]: float(probs_dl[i]) for i in range(len(CLASS_NAMES))}
        pred_lbl = CLASS_NAMES[int(np.argmax(probs_dl))]
        try:
            gradcam_rgb = compute_gradcam_overlay(c2_roi, LOADED_MODELS['mobilenetv2'])
        except Exception as err:
            gradcam_rgb = None
            print(f"[PERINGATAN] Grad-CAM gagal ({type(err).__name__}): {err}")
        preds['MobileNetV2'] = {
            'label': pred_lbl,
            'conf': prob_dict[pred_lbl] * 100.0,
            'probs': prob_dict,
            'ms': mnv2_ms
        }

    # 3. Prediksi YOLOv8n C2 No-HSV
    yolo_bbox = None
    if LOADED_MODELS['yolov8n'] is not None:
        t0 = time.perf_counter()
        y_out = LOADED_MODELS['yolov8n'].predict(c2_roi, imgsz=224, conf=0.15, verbose=False)[0]
        yolo_ms = (time.perf_counter() - t0) * 1000.0
        if len(y_out.boxes) > 0:
            best_i = int(y_out.boxes.conf.argmax().item())
            cls_id = int(y_out.boxes.cls[best_i].item())
            conf_y = float(y_out.boxes.conf[best_i].item()) * 100.0
            yolo_bbox = y_out.boxes.xyxy[best_i].cpu().numpy().astype(int).tolist()
            pred_lbl = IDX_TO_LABEL.get(cls_id, 'Unknown')
        else:
            pred_lbl = 'Tidak Terdeteksi'
            conf_y = 0.0
        preds['YOLOv8n'] = {
            'label': pred_lbl,
            'conf': conf_y,
            'bbox': yolo_bbox,
            'ms': yolo_ms
        }

    # 4. Konsensus Multi-Model (Majority Vote + Status QC)
    valid_votes = [(v['label'], v['conf']) for v in preds.values() if v['label'] in CLASS_NAMES]
    if valid_votes:
        counts = Counter([lbl for lbl, _ in valid_votes])
        top_count = counts.most_common(1)[0][1]
        top_labels = [lbl for lbl, cnt in counts.items() if cnt == top_count]
        if len(top_labels) == 1:
            consensus_lbl = top_labels[0]
        else:
            consensus_lbl = max(top_labels, key=lambda l: np.mean([c for lbl, c in valid_votes if lbl == l]))
        agree_str = f"{counts[consensus_lbl]}/{len(valid_votes)} model sepakat"
    else:
        consensus_lbl = 'N/A'
        agree_str = 'Tidak ada model aktif'
    if not pcd['qc_ok']:
        agree_str = f"QC GAGAL ({pcd['qc_msg']}); tidak andal"

    return {
        'filename': filename,
        'true_label': true_label,
        'pcd': pcd,
        'preds': preds,
        'gradcam_rgb': gradcam_rgb,
        'yolo_bbox': yolo_bbox,
        'consensus': consensus_lbl,
        'agreement': agree_str
    }

# === E. VISUALISASI DASHBOARD EKSEKUTIF 5 PANEL ===
def visualize_demo_result(res, save_path=None):
    pcd = res['pcd']
    preds = res['preds']

    fig, axes = plt.subplots(1, 5, figsize=(21.5, 4.5))

    # Panel 1: Foto Mentah Asli
    axes[0].imshow(cv2.cvtColor(pcd['im_work'], cv2.COLOR_BGR2RGB))
    ttl_1 = f"1. Input Mentah\n{res['filename']}"
    if res['true_label']:
        ttl_1 += f"\n[Aktual: {res['true_label']}]"
    axes[0].set_title(ttl_1, fontsize=10, fontweight='bold')
    axes[0].axis('off')

    # Panel 2: Mask Otsu b*+S + Morfologi
    axes[1].imshow(pcd['mask_full'], cmap='gray')
    axes[1].set_title(
        f"2. Otsu b*+S + LCC\nGamma={pcd['gamma_val']:.2f} | Area={pcd['leaf_area_pct']:.1f}%\nPCD: {pcd['pcd_ms']:.1f} ms",
        fontsize=10, fontweight='bold'
    )
    axes[1].axis('off')

    # Panel 3: C2 ROI 224x224 + Bounding Box YOLOv8n
    c2_disp = cv2.cvtColor(pcd['c2_roi'], cv2.COLOR_BGR2RGB).copy()
    if res['yolo_bbox'] is not None:
        x1, y1, x2, y2 = res['yolo_bbox']
        cv2.rectangle(c2_disp, (x1, y1), (x2, y2), (34, 197, 94), 2)
        y_info = preds['YOLOv8n']
        ttl_3 = f"3. C2 ROI + YOLOv8n\nPred: {y_info['label']} ({y_info['conf']:.1f}%)\nInfer: {y_info['ms']:.1f} ms"
    else:
        ttl_3 = "3. C2 ROI (224x224)\n(YOLOv8n nonaktif / 0 box)"
    axes[2].imshow(c2_disp)
    axes[2].set_title(ttl_3, fontsize=10, fontweight='bold')
    axes[2].axis('off')

    # Panel 4: Grad-CAM MobileNetV2
    if res['gradcam_rgb'] is not None:
        m_info = preds['MobileNetV2']
        axes[3].imshow(res['gradcam_rgb'])
        axes[3].set_title(
            f"4. Grad-CAM MobileNetV2\nPred: {m_info['label']} ({m_info['conf']:.1f}%)\nInfer: {m_info['ms']:.1f} ms",
            fontsize=10, fontweight='bold'
        )
    else:
        axes[3].imshow(cv2.cvtColor(pcd['c2_roi'], cv2.COLOR_BGR2RGB))
        axes[3].set_title("4. Grad-CAM MobileNetV2\n(tidak tersedia: belum dimuat / gagal)", fontsize=9.5)
    axes[3].axis('off')

    # Panel 5: Ringkasan Confidence, Latensi End-to-End, & Konsensus
    ax5 = axes[4]
    model_labels, conf_vals, bar_colors, ann_texts = [], [], [], []
    for m_key, nice_name in [('ML_Klasik', 'ML Klasik'), ('MobileNetV2', 'MobileNetV2'), ('YOLOv8n', 'YOLOv8n')]:
        if m_key in preds:
            p_lbl = preds[m_key]['label']
            p_cnf = preds[m_key]['conf']
            p_ms  = preds[m_key]['ms'] + pcd['pcd_ms']   # end-to-end (PCD + model)
            model_labels.append(nice_name)
            conf_vals.append(p_cnf)
            bar_colors.append(COLOR_MAP.get(p_lbl, '#64748b'))
            ann_texts.append(f"{p_lbl} ({p_cnf:.1f}% | {p_ms:.0f} ms)")

    y_pos = np.arange(len(model_labels))
    bars = ax5.barh(y_pos, conf_vals, color=bar_colors, edgecolor='black', height=0.52)
    ax5.set_yticks(y_pos)
    ax5.set_yticklabels(model_labels, fontsize=9.5, fontweight='bold')
    ax5.set_xlim(0, 148)
    ax5.set_xlabel('Confidence (%)', fontsize=9)
    ax5.invert_yaxis()
    ax5.grid(axis='x', alpha=0.3)
    for bar, txt in zip(bars, ann_texts):
        w = bar.get_width()
        x_txt = 4 if w >= 92 else w + 2
        txt_col = 'white' if w >= 92 else 'black'
        ax5.text(x_txt, bar.get_y() + bar.get_height()/2, txt,
                 va='center', ha='left', fontsize=8.2, fontweight='bold', color=txt_col)

    if not pcd['qc_ok']:
        cons_color = '#d97706'      # oranye: segmentasi diduga gagal
    elif res['true_label'] is None:
        cons_color = '#334155'      # abu netral: tanpa label aktual
    else:
        cons_color = '#16a34a' if res['consensus'] == res['true_label'] else '#dc2626'
    ax5.set_title(f"5. Konsensus: {res['consensus']}\n({res['agreement']})",
                  fontsize=9.2, fontweight='bold', color=cons_color)

    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=220, bbox_inches='tight')
    plt.show()

print("Pipeline PCD v2, QC Segmentasi, Ekstraksi 28 Fitur, & Mesin Inferensi siap dijalankan!")
```

## 2b. Uji Paritas Pipeline Demo vs Dataset v2 (Wajib Lulus Sebelum Uji Sesi 3)
Memverifikasi kesamaan piksel output `run_pcd_v2` terhadap `C2_pcd_masked` tersimpan dan kesamaan 28 fitur terhadap CSV fitur Skenario A Notebook 03, sekaligus mengaudit jumlah file di folder `Dataset` serta mencetak rentang area daun untuk kalibrasi `QC_AREA_LO/HI`.

```python
# === BAGIAN 2b: UJI PARITAS PIPELINE DEMO VS DATASET V2 (DENGAN TQDM & FILTER CSV) ===
from tqdm import tqdm

PARITY_SAMPLE_N = 30   # Uji cepat 30 sampel acak (ubah ke 127 untuk cek penuh seluruh dataset)

dfm = pd.read_csv(os.path.join(OUT_V2, 'dataset_master_p2.csv'))
orig = dfm[dfm['is_augmented'] == 0].reset_index(drop=True)
assert len(orig) == 127, f"dataset_master_p2.csv berisi {len(orig)} foto asli (seharusnya 127)."

# 1. Audit seluruh file di folder Dataset (instan, hanya baca nama file)
folder_label = {'P1_Sehat': 'Sehat', 'P1_Menguning': 'Menguning', 'P1_Kering': 'Kering'}
file_index = {}
for fol, lab_ in folder_label.items():
    for fp in glob.glob(os.path.join(RAW_DATASET_DIR, fol, '*')):
        if fp.lower().endswith(('.jpg', '.jpeg', '.png')):
            file_index.setdefault(os.path.basename(fp), []).append((fp, lab_))
n_files = sum(len(v) for v in file_index.values())
dups = {k: v for k, v in file_index.items() if len(v) > 1}
print(f"Foto di folder Dataset: {n_files} (dataset v2 = 127) | nama file ganda lintas folder: {len(dups)}")
for k, v in list(dups.items())[:12]:
    print("  ", k, "->", [l for _, l in v])

# 2. Paritas piksel C2 (demo vs file PNG tersimpan) + rentang area daun untuk QC
eval_orig = orig if (PARITY_SAMPLE_N is None or PARITY_SAMPLE_N >= len(orig)) else orig.sample(n=PARITY_SAMPLE_N, random_state=42)
rows = []
for _, r in tqdm(eval_orig.iterrows(), total=len(eval_orig), desc="Cek Paritas PCD v2"):
    cands = file_index.get(r['original_filename'], [])
    pick = [fp for fp, l in cands if l == r['label']] or [fp for fp, _ in cands]
    if not pick:
        rows.append({'sample_id': r['sample_id'], 'status': 'FILE HILANG'})
        continue
    im_raw = cv2.imread(pick[0])
    if im_raw is None:
        rows.append({'sample_id': r['sample_id'], 'status': 'GAGAL BACA'})
        continue
    out = run_pcd_v2(im_raw)
    ref = cv2.imread(r['path_C2_masked'])
    if ref is None or ref.shape != out['c2_roi'].shape:
        rows.append({'sample_id': r['sample_id'], 'status': 'UKURAN/FILE REF BERBEDA'})
        continue
    diff = np.abs(out['c2_roi'].astype(np.int16) - ref.astype(np.int16))
    rows.append({
        'sample_id': r['sample_id'], 'status': 'OK', 'label': r['label'], 'sesi': r['sesi'],
        'max_diff': int(diff.max()), 'mean_diff': float(diff.mean()),
        'gamma_abs_diff': abs(out['gamma_val'] - float(r['gamma_used'])),
        'area_pct': out['leaf_area_pct']
    })

df_par = pd.DataFrame(rows)
ok = df_par[df_par['status'] == 'OK']
print(f"\nPARITAS PIKSEL C2: {len(ok)}/{len(eval_orig)} foto terbandingkan | status lain: {df_par['status'].value_counts().to_dict()}")
if len(ok):
    print(f" - identik piksel (max_diff = 0) : {(ok['max_diff'] == 0).sum()} / {len(ok)}")
    print(f" - mean_diff terbesar            : {ok['mean_diff'].max():.4f}")
    print(f" - selisih gamma terbesar        : {ok['gamma_abs_diff'].max():.6f}")
    print(f" - area daun (%) min/p1/p99/max  : {ok['area_pct'].min():.1f} / {ok['area_pct'].quantile(.01):.1f} / {ok['area_pct'].quantile(.99):.1f} / {ok['area_pct'].max():.1f}")
    print("   -> sesuaikan QC_AREA_LO/HI di Bagian 2 dengan rentang ini (beri margin aman).")
    verdict = 'LULUS' if (len(ok) == len(eval_orig) and ok['mean_diff'].max() < 0.5) else 'GAGAL'
    print(f" >>> PARITAS PCD: {verdict} <<<")

# 3. Paritas 28 fitur vs CSV fitur Skenario A (PCD Masked) Notebook 03
feat_candidates = []
for p in glob.glob(os.path.join(OUT_V2, '**', '*.csv'), recursive=True):
    try:
        cols = set(pd.read_csv(p, nrows=1).columns)
    except Exception:
        continue
    if {'sample_id', 'H_mean', 'LBP_bin_9', 'GLCM_contrast'} <= cols:
        feat_candidates.append(p)

# Prioritaskan CSV fitur Skenario A (PCD Masked), hindari features_A0_raw atau C1
def _is_pcd_masked_csv(path):
    bn = os.path.basename(path).lower()
    return ('a0' not in bn) and ('raw' not in bn) and ('c1' not in bn) and ('nomask' not in bn)

feat_csv = next((p for p in sorted(feat_candidates) if _is_pcd_masked_csv(p)), feat_candidates[0] if feat_candidates else None)

if feat_csv is None:
    print("\n[INFO] CSV fitur PCD (Skenario A) Notebook 03 tidak ditemukan di OUT_V2.")
else:
    df_f = pd.read_csv(feat_csv).set_index('sample_id')
    worst = {}
    for _, r in orig.head(30).iterrows():
        if r['sample_id'] not in df_f.index:
            continue
        c2 = cv2.imread(r['path_C2_masked'])
        mk = cv2.imread(r['path_mask'], 0)
        f_demo = extract_handcrafted_features(c2, mask=mk, use_mask=True)
        for c in ALL_FEATURE_COLS:
            if c in df_f.columns:
                d = abs(f_demo[c] - float(df_f.loc[r['sample_id'], c])) / (abs(float(df_f.loc[r['sample_id'], c])) + 1e-9)
                worst[c] = max(worst.get(c, 0.0), d)
    bad = {c: v for c, v in worst.items() if v > 1e-3}
    print(f"\nPARITAS FITUR vs {os.path.basename(feat_csv)}: {len(worst)} fitur dicek pada 30 foto | melenceng > 0,1%: {len(bad)}")
    for c, v in sorted(bad.items(), key=lambda kv: -kv[1])[:10]:
        print(f"   {c:20s} selisih relatif maks = {v:.4f}")
    print(f" >>> PARITAS FITUR: {'LULUS' if not bad and len(worst) == 28 else 'PERIKSA'} <<<")
```

## 3. Mode 1: Uji Batch Folder Foto Daun Baru (Data Eksternal Sesi 3)
**Cara Penggunaan (jalankan setelah Bagian 2b lulus dan hash model dicatat):**
1. Foto daun kampus **BARU** (bukan daun Sesi 1 atau Sesi 2) di atas kertas putih polos: target awal **15 daun (5 per kelas)** untuk demo Progress 3, dan perluas menjadi **30 sampai 45 daun** untuk evaluasi Final. Jangan mencampur foto kondisi menyimpang (latar non-putih/paket stres) ke folder ini.
2. Unggah foto tersebut ke Google Drive pada folder `Project Akhir Computer Vision/Developments/Uji_Sesi3_Rabu/`.
   * **Opsi A (Dengan Label Aktual):** Buat subfolder `Sehat/`, `Menguning/`, dan `Kering/` lalu masukkan foto sesuai kondisinya agar Bagian 3b otomatis menghitung **Wilson CI 95% dan Confusion Matrix**.
   * **Opsi B (Tanpa Subfolder):** Taruh foto langsung di folder utama `Uji_Sesi3_Rabu/` untuk prediksi murni.
3. Klik **Run** pada sel di bawah ini. Jika folder masih kosong, skrip menjalankan *dry-run* pada 3 sampel latih (ditandai jelas sebagai *in-sample*, bukan akurasi eksternal).

```python
# === BAGIAN 3 (MODE 1): UJI BATCH FOLDER SESI 3 (DRIVE) + AUTO DRY-RUN ===
from sklearn.metrics import accuracy_score, f1_score

DRY_RUN_IF_EMPTY = True   # Otomatis uji 3 sampel dataset jika Uji_Sesi3_Rabu masih kosong
MAX_PLOTS_SHOWN  = 10      # Maksimal jumlah dashboard 5 panel yang ditampilkan ke layar

def collect_test_images(root_dir):
    items = []
    has_subfolders = any(os.path.isdir(os.path.join(root_dir, c)) for c in CLASS_NAMES)
    if has_subfolders:
        for c in CLASS_NAMES:
            c_dir = os.path.join(root_dir, c)
            if not os.path.isdir(c_dir):
                continue
            for fp in sorted(glob.glob(os.path.join(c_dir, '*'))):
                if fp.lower().endswith(('.jpg', '.jpeg', '.png')):
                    items.append((fp, c))
    for fp in sorted(glob.glob(os.path.join(root_dir, '*'))):
        if os.path.isfile(fp) and fp.lower().endswith(('.jpg', '.jpeg', '.png')):
            items.append((fp, None))
    return items

test_items = collect_test_images(TEST_SESI3_DIR)
is_dry_run = False

if len(test_items) == 0 and DRY_RUN_IF_EMPTY:
    is_dry_run = True
    print(f"[INFO DRY-RUN] Folder '{TEST_SESI3_DIR}' masih kosong.")
    print("Mengambil 3 contoh foto (1 Sehat, 1 Menguning, 1 Kering) dari folder Dataset untuk verifikasi alur demo...")
    print("PERHATIAN: foto ini adalah DATA LATIH (IN-SAMPLE). Hanya uji alur, BUKAN akurasi eksternal.\n")
    for folder_name, lbl in [('P1_Sehat', 'Sehat'), ('P1_Menguning', 'Menguning'), ('P1_Kering', 'Kering')]:
        f_list = sorted([
            p for p in glob.glob(os.path.join(RAW_DATASET_DIR, folder_name, '*'))
            if p.lower().endswith(('.jpg', '.jpeg', '.png'))
        ])
        if f_list:
            test_items.append((f_list[-1], lbl))

if len(test_items) == 0:
    print("Tidak ada foto ditemukan. Silakan unggah foto daun baru ke folder Uji_Sesi3_Rabu.")
else:
    mode_tag = "DRY-RUN" if is_dry_run else "UJI EKSTERNAL SESI 3"
    print(f"=== MEMULAI {mode_tag} ({len(test_items)} FOTO) ===")
    summary_rows = []
    for i, (fpath, true_lbl) in enumerate(test_items):
        fname = os.path.basename(fpath)
        img_bgr = cv2.imread(fpath)
        if img_bgr is None:
            print(f"[PERINGATAN] Gagal membaca gambar: {fname}")
            continue

        res = run_multimodel_inference(img_bgr, filename=fname, true_label=true_lbl)
        p = res['preds']
        row = {
            'Filename': fname,
            'Aktual': true_lbl if true_lbl else '-',
            'PCD_ms': round(res['pcd']['pcd_ms'], 1),
            'Gamma': round(res['pcd']['gamma_val'], 2),
            'Area_pct': round(res['pcd']['leaf_area_pct'], 1),
            'QC': res['pcd']['qc_msg']
        }
        for m_key in ['ML_Klasik', 'MobileNetV2', 'YOLOv8n']:
            if m_key in p:
                row[f'{m_key}_Pred'] = p[m_key]['label']
                row[f'{m_key}_Conf'] = round(p[m_key]['conf'], 1)
                row[f'{m_key}_ms']   = round(p[m_key]['ms'], 1)
        row['Konsensus']   = res['consensus']
        row['Kesepakatan'] = res['agreement']
        summary_rows.append(row)

        if i < MAX_PLOTS_SHOWN:
            save_fig = os.path.join(DEMO_OUT_DIR, f"Demo_{i+1:02d}_{os.path.splitext(fname)[0]}.png")
            visualize_demo_result(res, save_path=save_fig)

    df_demo = pd.DataFrame(summary_rows)
    csv_out = os.path.join(DEMO_OUT_DIR, 'Hasil_Demo_DryRun.csv' if is_dry_run else 'Hasil_Uji_Eksternal_Sesi3.csv')
    df_demo.to_csv(csv_out, index=False)

    n_qc_bad = int((df_demo['QC'] != 'OK').sum())
    print(f"\nQC segmentasi: {n_qc_bad} dari {len(df_demo)} foto ditandai tidak andal.")
    print(f"\n=== TABEL RINGKASAN PREDIKSI ({mode_tag}) ===")
    print(df_demo.to_string(index=False))

    labeled_df = df_demo[df_demo['Aktual'].isin(CLASS_NAMES)]
    if len(labeled_df) > 0:
        tag_ = " (IN-SAMPLE, BUKAN AKURASI EKSTERNAL)" if is_dry_run else ""
        print(f"\n=== REKAPITULASI AKURASI & LATENSI RATA-RATA (N = {len(labeled_df)} BERLABEL){tag_} ===")
        for col_pred, col_ms, nice in [
            ('ML_Klasik_Pred', 'ML_Klasik_ms', 'Classical ML (28 Fitur)'),
            ('MobileNetV2_Pred', 'MobileNetV2_ms', 'MobileNetV2 (C2 Masked)'),
            ('YOLOv8n_Pred', 'YOLOv8n_ms', 'YOLOv8n (C2 No-HSV)'),
            ('Konsensus', None, 'Konsensus Multi-Model')
        ]:
            if col_pred not in labeled_df.columns:
                continue
            acc = accuracy_score(labeled_df['Aktual'], labeled_df[col_pred]) * 100.0
            f1  = f1_score(labeled_df['Aktual'], labeled_df[col_pred], average='macro', zero_division=0) * 100.0
            ms_txt = f" | Avg Infer: {labeled_df[col_ms].mean():.2f} ms" if (col_ms and col_ms in labeled_df.columns) else ""
            print(f" - {nice:25s}: Accuracy = {acc:6.2f}% | Macro-F1 = {f1:6.2f}%{ms_txt}")
        print(f" - Rata-rata waktu PCD v2   : {labeled_df['PCD_ms'].mean():.2f} ms/citra")

    print(f"\nTabel dan visualisasi hasil tersimpan di: {DEMO_OUT_DIR}")
```

## 3b. Laporan Evaluasi Eksternal (CI Wilson 95%, Confusion Matrix, & Recall per Kelas)
Menghitung statistik evaluasi lengkap khusus data eksternal Sesi 3 (otomatis menolak pelaporan metrik apabila masih dalam mode *dry-run* in-sample).

```python
# === BAGIAN 3b: LAPORAN EVALUASI EKSTERNAL (CI WILSON 95%, CONFUSION MATRIX, RECALL PER KELAS) ===
from sklearn.metrics import confusion_matrix, f1_score, recall_score
from math import sqrt

def wilson(k, n, z=1.96):
    p = k / n
    d = 1 + z * z / n
    c = p + z * z / (2 * n)
    h = z * sqrt(p * (1 - p) / n + z * z / (4 * n * n))
    return (c - h) / d * 100, (c + h) / d * 100

assert 'df_demo' in globals(), "Jalankan sel Mode 1 (Bagian 3) terlebih dahulu."
lab_df = df_demo[df_demo['Aktual'].isin(CLASS_NAMES)].copy()

if is_dry_run:
    print("DRY-RUN: foto berasal dari dataset latih (IN-SAMPLE). Angka ini hanya uji alur, JANGAN dilaporkan sebagai akurasi eksternal.")
elif len(lab_df) == 0:
    print("Belum ada foto berlabel (taruh foto di subfolder Sehat/, Menguning/, Kering/).")
else:
    n = len(lab_df)
    print(f"N foto berlabel = {n} | sebaran: {lab_df['Aktual'].value_counts().to_dict()}")
    print(f"Baseline kelas mayoritas = {lab_df['Aktual'].value_counts().max() / n * 100:.1f}% | tebakan acak = 33,3%\n")
    model_cols = [
        ('ML_Klasik_Pred', 'ML Klasik'), ('MobileNetV2_Pred', 'MobileNetV2'),
        ('YOLOv8n_Pred', 'YOLOv8n'), ('Konsensus', 'Konsensus')
    ]
    model_cols = [(c, nm) for c, nm in model_cols if c in lab_df.columns]
    fig, axes = plt.subplots(1, len(model_cols), figsize=(4.6 * len(model_cols), 4.2))
    axes = np.atleast_1d(axes)
    rows = []
    for ax, (col, nm) in zip(axes, model_cols):
        y, p = lab_df['Aktual'], lab_df[col]
        k = int((y == p).sum())
        lo, hi = wilson(k, n)
        f1 = f1_score(y, p, labels=CLASS_NAMES, average='macro', zero_division=0) * 100
        rec = recall_score(y, p, labels=CLASS_NAMES, average=None, zero_division=0) * 100
        rows.append({
            'Model': nm, 'N': n, 'Benar': k, 'Akurasi (%)': round(k / n * 100, 1),
            'CI95 Wilson': f"{lo:.0f} s.d. {hi:.0f}%", 'Macro-F1 (%)': round(f1, 1),
            'Recall Sehat': round(rec[0], 1), 'Recall Menguning': round(rec[1], 1), 'Recall Kering': round(rec[2], 1)
        })
        cols = CLASS_NAMES + ['Tidak Terdeteksi']
        cm = confusion_matrix(y, p, labels=cols)[:3, :]
        ax.imshow(cm, cmap='Blues')
        ax.set_xticks(range(4)); ax.set_xticklabels(['Sehat', 'Mengun.', 'Kering', 'N/A'], fontsize=9)
        ax.set_yticks(range(3)); ax.set_yticklabels(CLASS_NAMES, fontsize=9)
        for i in range(3):
            for j in range(4):
                ax.text(j, i, int(cm[i, j]), ha='center', va='center', fontsize=13, fontweight='bold',
                        color='white' if cm[i, j] > cm.max() / 2 else 'black')
        ax.set_title(f"{nm}\nAkurasi {k}/{n} = {k/n*100:.1f}% (CI {lo:.0f} s.d. {hi:.0f}%)", fontsize=9.8, fontweight='bold')
        ax.set_xlabel('Prediksi'); ax.set_ylabel('Aktual')
    plt.tight_layout()
    plt.savefig(os.path.join(DEMO_OUT_DIR, 'Sesi3_Confusion_Matrix.png'), dpi=220, bbox_inches='tight')
    plt.show()
    df_ext = pd.DataFrame(rows)
    print(df_ext.to_string(index=False))
    df_ext.to_csv(os.path.join(DEMO_OUT_DIR, 'Sesi3_Ringkasan_Evaluasi.csv'), index=False)
    print("\nCatatan: CI lebar pada N kecil; selisih tipis antar-model belum tentu signifikan secara statistik.")
```

```python
# === BAGIAN 5 (MODE 3): LIVE CAPTURE WEBCAM LAPTOP DI GOOGLE COLAB ===
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

def take_photo_webcam(filename='/content/webcam_leaf.jpg', quality=0.92):
    js = Javascript('''
        async function takePhoto(quality) {
            const div = document.createElement('div');
            const capture = document.createElement('button');
            capture.textContent = 'Ambil Foto Daun';
            capture.style.padding = '8px 18px';
            capture.style.fontSize = '15px';
            capture.style.fontWeight = 'bold';
            capture.style.marginBottom = '8px';
            div.appendChild(capture);

            const video = document.createElement('video');
            video.style.display = 'block';
            video.style.maxWidth = '560px';
            const stream = await navigator.mediaDevices.getUserMedia({video: true});
            document.body.appendChild(div);
            div.appendChild(video);
            video.srcObject = stream;
            await video.play();

            google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);
            await new Promise((resolve) => capture.onclick = resolve);

            const canvas = document.createElement('canvas');
            canvas.width = video.videoWidth;
            canvas.height = video.videoHeight;
            canvas.getContext('2d').drawImage(video, 0, 0);
            stream.getVideoTracks()[0].stop();
            div.remove();
            return canvas.toDataURL('image/jpeg', quality);
        }
    ''')
    display(js)
    data = eval_js(f'takePhoto({quality})')
    binary = b64decode(data.split(',')[1])
    with open(filename, 'wb') as f:
        f.write(binary)
    return filename

try:
    cam_path = take_photo_webcam('/content/webcam_leaf.jpg')
    img_cam = cv2.imread(cam_path)
    res_cam = run_multimodel_inference(img_cam, filename='webcam_leaf.jpg', true_label=None)
    save_cam = os.path.join(DEMO_OUT_DIR, f"Webcam_{int(time.time())}.png")
    visualize_demo_result(res_cam, save_path=save_cam)
    print(f"\n>>> KONSENSUS WEBCAM: {res_cam['consensus']} ({res_cam['agreement']}) <<<")
    print(f"QC segmentasi: {res_cam['pcd']['qc_msg']} | catatan: domain webcam berbeda dari foto HP, tidak dihitung sebagai bukti performa.")
except Exception as err:
    print("Gagal mengakses webcam (pastikan izin kamera browser aktif):", err)
```
